```python
# Program to calculate total marks and percentage of a student

# Input student details
name = input("Enter student name: ")
student_class = input("Enter class: ")

# Input marks for 5 subjects
mark1 = float(input("Enter marks for Subject 1: "))
mark2 = float(input("Enter marks for Subject 2: "))
mark3 = float(input("Enter marks for Subject 3: "))
mark4 = float(input("Enter marks for Subject 4: "))
mark5 = float(input("Enter marks for Subject 5: "))

# Calculate total and percentage
total = mark1 + mark2 + mark3 + mark4 + mark5
percentage = (total / 500) * 100

# Display result
print("Student Result ")
print("Name:", name)
print("Class:", student_class)
print("Total Marks:", total)
print("Percentage:", percentage, "%")
```

```python
#  Input 2 strings concatinate them and store in another variable. then perform generally used string methods on it like
# lower(), upper(), title(), swapcase(), capitalize(), casefold(), center(), count(), endswith(), find(), isalnum(), isdigit(), isnumeric(), isspace(), replace()

# Input two strings
string1 = input("Enter first string: ")
string2 = input("Enter second string: ")

# Concatenate the strings
concatenated_string = string1 + string2

# Perform string methods
print("Lowercase:", concatenated_string.lower())
print("Uppercase:", concatenated_string.upper())
print("Title Case:", concatenated_string.title())
print("Swapcase:", concatenated_string.swapcase())
print("Capitalized:", concatenated_string.capitalize())
print("Casefold:", concatenated_string.casefold())
print("Centered:", concatenated_string.center(50))
print("Count of 'a':", concatenated_string.count('a'))
print("Ends with 'world':", concatenated_string.endswith("world"))
print("Find 'hello':", concatenated_string.find("hello"))
print("Is alphanumeric:", concatenated_string.isalnum())
print("Is digit:", concatenated_string.isdigit())
print("Is numeric:", concatenated_string.isnumeric())
print("Is space:", concatenated_string.isspace())
print("Replace 'hello' with 'hi':", concatenated_string.replace("hello", "hi"))
```

```python
#In your last program where you find the total and percentage of a student's marks of 5 subject, find the grade of the student using conditional statement. Eg. grade 'A' if percentage is greator than or equals to 60, 'B' for  percentage is greator than or equals to 50 and less than 60,  'C' for  percentage is greator than or equals to 40 and less than 50,  'D' for  percentage is greator than or equals to 33 and less than 40, otherwise 'Fail'

# Input student details
name = input("Enter student name: ")
student_class = input("Enter class: ")

# Input marks for 5 subjects
mark1 = float(input("Enter marks for Subject 1: "))
mark2 = float(input("Enter marks for Subject 2: "))
mark3 = float(input("Enter marks for Subject 3: "))
mark4 = float(input("Enter marks for Subject 4: "))
mark5 = float(input("Enter marks for Subject 5: "))
# Calculate total and percentage
total = mark1 + mark2 + mark3 + mark4 + mark5   
percentage = (total / 500) * 100

# Determine grade
if percentage >= 60:
    grade = 'A'
elif percentage >= 50:
    grade = 'B'
elif percentage >= 40:
    grade = 'C'
elif percentage >= 33:
    grade = 'D'
else:
    grade = 'Fail'
# Display result
print("Student Result ")
print("Name:", name)
print("Class:", student_class)
print("Total Marks:", total)
print("Percentage:", percentage, "%")
print("Grade:", grade)
```
