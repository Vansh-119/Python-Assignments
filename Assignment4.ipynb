```python
#Q1
def maximum(a, b, c):
    return max(a, b, c)

print("Maximum:", maximum(10, 25, 15))
```

```python
#Q2
def distinct_list(lst):
    return list(set(lst))

numbers = [1, 2, 2, 3, 4, 4, 5]
print("Distinct List:", distinct_list(numbers))
```

```python
#Q3
def multiply_list(lst):
    result = 1
    for num in lst:
        result *= num
    return result

numbers = [2, 3, 4]
print("Multiplication:", multiply_list(numbers))
```

```python
#Q4
def factorial(n):
    if n == 0 or n == 1:
        return 1
    fact = 1
    for i in range(2, n + 1):
        fact *= i
    return fact

print("Factorial:", factorial(5))
```

```python
#Q5
def reverse_string(s):
    return s[::-1]

text = input("Enter a string: ")
print("Reversed String:", reverse_string(text))
```

```python
#Q6
def check_range(num, start, end):
    if start <= num <= end:
        print(num, "is in the range")
    else:
        print(num, "is not in the range")

check_range(5, 1, 10)
```

```python
#Q7
def even_numbers(lst):
    for num in lst:
        if num % 2 == 0:
            print(num)

numbers = [1, 2, 3, 4, 5, 6]

even_numbers(numbers)
```

```python
#Q8
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

num = int(input("Enter a number: "))
if is_prime(num):
    print(num, "is Prime")
else:
    print(num, "is not Prime")
```

```python
#Q9
def count_case(text):
    upper = 0
    lower = 0

    for char in text:
        if char.isupper():
            upper += 1
        elif char.islower():
            lower += 1

    print("Upper case letters:", upper)
    print("Lower case letters:", lower)

text = input("Enter a string: ")
count_case(text)
```
